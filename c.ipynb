ΕΡΩΤΗΜΑ Γ

```python
import pandas as pd
```

```python
full_set = pd.concat(
    [
        pd.read_csv("TrainingSet.csv"),
        pd.read_csv("TestSet.csv")
    ],
    ignore_index=True
)

sold_set = pd.concat(
    [
        pd.read_csv("TrainingSubset.csv"),
        pd.read_csv("TestSubset.csv")
    ],
    ignore_index=True
)

print(full_set.shape, sold_set.shape)
```

```python
sold_set.head()
```

Ορισμός στόχου και συνόλου features.

```python
y = (sold_set["Price"] >= sold_set["AvgPrice"]).astype(int)
X = sold_set.drop(columns = ["Price","AvgPrice", "EbayID", "PersonID", "SellerName", "PricePercent"])
```

```python
print(y.value_counts())
```

Εκπαίδευση

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state= 42, stratify = y)
```

ιαχωρισμός numeric με categorical τιμές

```python
# Numeric features
numeric_features = X_train.select_dtypes(include=["int64", "float64"]).columns

# Categorical features
categorical_features = X_train.select_dtypes(include=["object"]).columns

print("Numeric features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
```

```python
# Data Preproccesing

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers= [
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown= "ignore"), categorical_features)

    ]
)
```

Αλγόριθμος 1: Logistic Regression

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

model_LR = Pipeline(
    steps=[
        ("preprocess", preprocessor),
        ("model", LogisticRegression(max_iter=2000))
    ]
)
```

```python
model_LR.fit(X_train, y_train)
```

```python
y_pred = model_LR.predict(X_test)
```

Αλγόριθμος 2: linear SVM

```python
from sklearn.svm import LinearSVC

model_SVM = Pipeline(
    steps=[
        ("preprocess", preprocessor),
        ("model", LinearSVC(
            C= 0.1,
            max_iter=2000))
    ]
)
```

```python
model_SVM.fit(X_train, y_train)
y_pred_svm = model_SVM.predict(X_test)
```

```python
test_accLR = model_LR.score(X_test, y_test)
train_accLR = model_LR.score(X_train, y_train)

print("\nTrain accuracy LR:", train_accLR)
print("\nTest accuracy LR:", test_accLR)

test_accSVM = model_SVM.score(X_test, y_test)
train_accSVM = model_SVM.score(X_train, y_train)

print("\nTrain accuracy SVM:", train_accSVM)
print("\nTest accuracy SVM:", test_accSVM,"\n\n")
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print("\nClassification Report (LR)\n",classification_report(y_test, y_pred, digits=4))
print("\nConfusion matrix (Logisitic regression):\n", confusion_matrix(y_test, y_pred))
```

```python
print("\nClassification Report (SVM)\n",classification_report(y_test, y_pred_svm, digits = 4))
print("\nConfusion matrix (SVM):\n", confusion_matrix(y_test, y_pred_svm))
```
