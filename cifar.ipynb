Φόρτωση dataset με βάση το documentation

```python
import pickle

filepath = "cifar-10-batches-py"
def load_cifar_batch(filepath):
    with open(filepath, 'rb') as f:
        batch = pickle.load(f, encoding='bytes')
    return batch
```

Φόρτωση κάθε ενός batch αρχείου από τον φάκελο και χωρίζω σε train, test

Χρήση array της numpy

```python
import numpy as np

X_train_list = []
y_train_list = []

for i in range(1, 6):
    batch = load_cifar_batch(f"cifar-10-batches-py/data_batch_{i}")
    X_train_list.append(batch[b'data'])
    y_train_list.extend(batch[b'labels'])

#ενοποίηση arrays κάθετα
X_train = np.vstack(X_train_list)
y_train = np.array(y_train_list)
```

```python
meta = load_cifar_batch("cifar-10-batches-py/batches.meta")
label_names = meta[b'label_names']
print(label_names)
```

Φόρτωση και δημιουργία test data

```python
test_batch = load_cifar_batch("cifar-10-batches-py/test_batch")

X_test = test_batch[b'data']
y_test = test_batch[b'labels']
```

```python
print(X_test)
```

```python
print(y_test[:50])
```

Επιλογή υπό-κλάσεων

```python
selected_classes = [2,3,5]
classes_names = ['bird', 'cat', 'dog']

y_test = np.array(y_test)


mask_train = np.isin(y_train, selected_classes)
X_train_sub = X_train[mask_train]
y_train_sub = y_train[mask_train]

mask_test = np.isin(y_test, selected_classes)
X_test_sub = X_test[mask_test]
y_test_sub = y_test[mask_test]
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

Εφαρμογή Scaling στα δεδομένα για πιο αξιόλογα αποτελέσματα.

```python
scaler = StandardScaler()

scaler.fit(X_train_sub)

X_test_scaled = scaler.transform(X_test_sub)
X_train_scaled = scaler.transform(X_train_sub)
```

Εφαρμογή PCA.

Μείωση των components σε 150.

```python
pca = PCA(n_components= 150)

pca.fit(X_train_scaled)

X_test = pca.transform(X_test_scaled)
X_train = pca.transform(X_train_scaled)
```

Βλέπουμε όντως ότι τα features μειώθηκαν σε 150.

```python
print(X_train.shape)
```

Μοντέλο 1: Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression(max_iter= 2000)

log_reg.fit(X_train, y_train_sub)
```

```python
log_reg_pred = log_reg.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report

log_reg_cl_rep = classification_report(y_test_sub, log_reg_pred, digits= 4)
print("\nClassification Report LR:\n",log_reg_cl_rep)
```

```python
log_reg_conf_matrix = confusion_matrix(y_test_sub, log_reg_pred)
print("\nConfusion Matrix LR\n\n",log_reg_conf_matrix)
```

```python
test_accLR = log_reg.score(X_test, y_test_sub)
train_accLR = log_reg.score(X_train, y_train_sub)

print("\nTrain accuracy LR:", train_accLR)
print("\nTest accuracy LR:", test_accLR)
```

Μοντέλο 2: KNN as classifier

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors= 10)
```

```python
knn.fit(X_train, y_train_sub)
```

```python
knn_pred = knn.predict(X_test)
```

```python
test_accKNN = knn.score(X_test, y_test_sub)
train_accKNN = knn.score(X_train, y_train_sub)

print("\nTrain accuracy KNN:", train_accKNN)
print("\nTest accuracy KNN:", test_accKNN)
```

```python
knn_cl_rep = classification_report(y_test_sub, knn_pred, digits= 4)
print("\nClassification Report KNN:\n",knn_cl_rep)
```

```python
print("\nConfusion Matrix KNN:\n\n", confusion_matrix(y_test_sub, knn_pred))
```
